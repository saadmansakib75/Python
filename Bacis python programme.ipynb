#1.Simple Calculator Program

```python
num1 = float(input("Enter first number:"))   #get input 1st number
operator = input("Enter operator:")          #get input operator
num2 = float(input("Enter second number:"))  #get input @nd number

if operator == '+':                          #program for addion
    result = num1+num2
elif operator =="-":                         #program for subtraction
    result = num1-num2
elif operator == "*":                        #program for multiplication
    result = num1*num2
elif operator == "/":                        #program for devition
    result = num1/num2
else :
    result = 'invalid'


print(f'Result :{result}')                      #Showing output the result



```
