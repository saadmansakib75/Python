#3.String Manipulation

```python
string = input("Enter a string: ")
print(f"String in uppercase: {string.upper()}")
print(f"Length of the string: {len(string)}")
print(f"String in reverse: {string[:: -1]}")
```

#1.String Processing with Loop

```python
text = input("Enter a text: ")
count = 0
for i in text:
  if i.lower() in "aeiou":
    count +=1

print(f"Number of vowle in the string: {count}")

```

#5.List Comprehension

```python
numbers = [10,20,30,40,50,60,70]

new_list = []


for i in numbers:
  if i >= 40:
    new_list.append(i)


print(new_list)

```
